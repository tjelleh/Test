```python
from ROOT import gStyle, TCanvas, TH1D, TLegend, kRed, kBlue
gStyle.SetOptStat(0)#Set this so that the stats box is not shown on the plots
#Create two histograms with 50 bins ranging from -3 to 3 with two different

hist1 = TH1D("Gauss-Kurve", "Gauss x Polynomial", 50, -3, 3)

hist2 = TH1D("Polynom-Kurve", "Polynomial", 50, -3, 3)

#Fill first histogram with Gaussian distribution with 10000 entries.
#Fill second histogram with a second order polynomial and 5000 entries
hist1.FillRandom("gaus", 10000)
hist2.FillRandom("pol2", 5000)

# Achsen zuweisen
hist1.GetXaxis().SetTitle("Wertebereich")
hist1.GetYaxis().SetTitle("Einträge pro Bin")

# --- HIER DEN ABSTAND ANPASSEN ---
hist1.GetXaxis().SetTitleOffset(1.3)  
hist1.GetYaxis().SetTitleOffset(1.3)  

# Achsen zentrieren
hist1.GetXaxis().CenterTitle(True)  
hist1.GetYaxis().CenterTitle(True)


#Set the line color of the first histogram to kRed and second to kBlue.
hist1.SetLineColor(kRed)
hist2.SetLineColor(kBlue)

#FIll Color einstellen
hist1.SetFillColorAlpha(kRed, 0.3)
hist2.SetFillColorAlpha(kBlue, 0.3)


#Draw both histograms on a canvas.
canvas = TCanvas()
hist1.Draw("hist f")
hist2.Draw("hist f same")

#canvas.Draw()
#Clone both histograms and normalise them (scale with inverse of the integral).
hist1_norm = hist1.Clone("hist1_norm")
hist1_norm.Scale(1 / hist1.Integral())
hist2_norm = hist2.Clone("hist2_norm")
hist2_norm.Scale(1 / hist2.Integral())

hist2_norm.SetFillStyle(0)
hist1_norm.SetFillStyle(0)
#Draw both histograms on a different canvas.
canvas2 = TCanvas()
hist1_norm.Draw("hist")
hist2_norm.Draw("hist same")
#canvas2.Draw()
#Draw a legend on both canvases at position (0.7, 0.75, 0.90, 0.87)
legend = TLegend(0.7, 0.75, 0.9, 0.87)
legend.AddEntry(hist1, "Gauss", "l")
legend.AddEntry(hist2, "Polynomial", "l")
legend.SetBorderSize(2)


canvas.cd()
legend.Draw()
canvas2.cd()
legend.Draw()
canvas.Draw()
canvas2.Draw()
#Write the output to a file
#canvas.SaveAs("ExerciseCanvas.pdf")
#canvas2.SaveAs("ExerciseCanvas2.pdf")
```
